### Target variable: class balance
Attrition is imbalanced (far fewer leavers than stayers), so rates matter more than raw counts.

```python
counts = raw['Attrition'].value_counts()
display(counts.to_frame('employees').assign(pct=lambda d: (100 * d.employees / d.employees.sum()).round(2)))
```

### Observations
- No missing values or duplicate rows.
- `EmployeeCount`, `Over18`, `StandardHours` are constants and carry no information.
- Ordinal survey columns (satisfaction, involvement, ...) are stored as integers 1-4; they are categories, not continuous measures.
