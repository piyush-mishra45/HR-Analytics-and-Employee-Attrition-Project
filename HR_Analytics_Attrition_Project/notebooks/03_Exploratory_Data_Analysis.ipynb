# 03 - Exploratory Data Analysis
Attrition patterns across the workforce. Charts are also saved to `reports/figures/`.

```python
import sys
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display, Image

# Locate the project root whether the notebook is run from notebooks/ or the root
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT / "python"))

import config as cfg
import data_cleaning as dc
import eda_analysis as eda
import visualization as viz

pd.set_option("display.max_columns", 50)
viz.set_theme()
```

```python
df = eda.load_clean()
T = eda.summary_tables(df)
overall = eda.overall_attrition(df)
company_rate = overall['attrition_rate_pct']
overall
```
